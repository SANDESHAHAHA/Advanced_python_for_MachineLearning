```python
new_df = pd.DataFrame({
    'Name':['Ram',"Shyam",'Hari'],
    'Math':[23,435,56],
    'Science':[34,45,56],
    "Physics":[34,45,56]
})
new_df
```

```python
long_form = pd.melt(
    new_df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Scores'
)
long_form
```

```python
wide_format = long_form.pivot_table(
    index='Name',
    columns='Subject',
    values='Scores',
)
wide_format
```

```python
df = pd.DataFrame({
    'Department':['IT','IT','HR','HR'],
    'Employee':['A','B','C','D'],
    'Salary':[6000,5000,3000,5000]
})
df.head()
```

```python
table = df.pivot_table(
   index='Department',
   values='Salary',
   aggfunc='mean'
)
table
```

```python
df['dept_mean'] = df.groupby('Department')['Salary'].transform('mean')
```

```python
df_categorical = pd.DataFrame({
    'id':[1,2,3],
    'Name':['Ram',"Shyam","Hari"],
    'Department':["IT","HR","PR"]
})

df_categorical
```

```python
df_categorical.info()
```
